Como mostrar a distribuição dos estudantes entre turmas sem induzir uma comparação enganosa?

## Conceitos

Frequência absoluta é a quantidade de observações em uma categoria. Frequência relativa divide essa contagem pelo total considerado. Uma porcentagem multiplica a proporção por 100. Declare o denominador, especialmente quando houver ausências.

Gráfico de barras compara categorias; histograma agrupa valores quantitativos em intervalos e mostra sua distribuição. Intervalos diferentes podem mudar a aparência de um histograma. Uma categoria numérica como ID continua sendo categoria, não uma escala para histograma.

Um gráfico precisa de título, rótulos, unidades e uma escala consistente. Em barras de contagem, iniciar o eixo vertical em zero permite comparar comprimentos. Para comparar distribuições, mantenha escalas e intervalos iguais entre grupos.

## Exemplo resolvido
Antes de executar, preveja a saída.

```python
import pandas as pd
import matplotlib.pyplot as plt
turmas = pd.Series(["A", "A", "A", "B"] )
f = turmas.value_counts().sort_index()
print(f)
print(f / f.sum() * 100)
f.plot.bar(title="Estudantes sintéticos por turma", ylabel="Estudantes", xlabel="Turma")
plt.tight_layout()
plt.show()
```

A tem três estudantes (75%) e B um (25%). Essa distribuição descreve os quatro registros, sem representar necessariamente a distribuição de uma escola.
