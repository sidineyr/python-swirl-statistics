Uma nota está em branco. Ela deve ser substituída por zero antes de comparar as turmas?

## Conceitos

Um valor ausente indica que não temos uma observação utilizável naquele campo. Pode resultar de não resposta, falha de registro ou não aplicabilidade. Zero é um valor observado e tem significado próprio; substituir ausência por zero muda esse significado.

isna() identifica ausências e sum() conta True como 1. dropna(subset=["nota"]) remove somente linhas com ausência em nota. Não descarte linhas por ausências em campos irrelevantes para sua pergunta. Preserve o original em outra variável ou arquivo.

Preencher com média ou mediana é uma decisão analítica, não uma correção automática. Pode reduzir artificialmente a variabilidade. Excluir também pode introduzir viés se a ausência estiver relacionada ao fenômeno. Informe quantos registros entraram e saíram e por quê.

Duplicata exata e pessoa duplicada não são a mesma coisa. Um estudante pode ter várias avaliações legítimas. Valide o identificador e a regra de coleta antes de remover registros. Valores extremos exigem investigação, não exclusão por aparência.

```python
def conferir(valor, esperado, explicacao, tolerancia=0.01):
    """Aceita qualquer procedimento que produza um resultado numérico equivalente."""
    if isinstance(valor, bool) or not isinstance(valor, (int, float)):
        print("Tipo de resposta: informe um número, não texto ou True/False.")
        return False
    if not math.isfinite(valor) or not math.isclose(valor, esperado, abs_tol=tolerancia, rel_tol=1e-9):
        print("Cálculo: resultado diferente do esperado. Revise os dados e o denominador.")
        return False
    print("Resultado conferido. " + explicacao)
    return True
```

## Exemplo resolvido
Antes de executar, preveja a saída.

```python
import pandas as pd
df = pd.DataFrame({"nota": [0, 8, None, 10]})
print(df["nota"].isna().sum())
validas = df.dropna(subset=["nota"]).copy()
print(validas["nota"].mean())
```

Há uma ausência. A média das três notas observadas é 6. O zero permanece. Se a ausência fosse preenchida com zero, a média cairia para 4.5: outra suposição, outro resultado.
