Os estudantes têm identificadores numéricos, turmas e notas. Faz sentido calcular a média dos identificadores?

## Conceitos

Uma variável qualitativa representa categorias. Nominal não tem ordem intrínseca, como turma A/B; ordinal tem ordem, como satisfação baixa/média/alta, sem garantir distâncias iguais entre níveis.

Uma variável quantitativa representa uma quantidade. Discreta resulta de contagens, como número de faltas; contínua representa medidas em uma escala, como duração. Uma medida registrada em minutos inteiros pode ser discretizada pelo instrumento sem que a duração deixe de ser contínua.

Um código escrito com dígitos pode continuar sendo nominal: estudante 101 e estudante 102 são identificadores. Uma média desses códigos não descreve uma quantidade substantiva. Para categorias use frequências; para quantidades escolha medidas de centro e dispersão considerando o contexto.

No Python, uma coluna object pode guardar categorias ou texto numérico. O significado vem da coleta e do dicionário de dados. Leia as unidades e a forma de registro antes de escolher operações.

```python
def conferir(valor, esperado, explicacao, tolerancia=0.01):
    """Aceita qualquer procedimento que produza um resultado numérico equivalente."""
    if isinstance(valor, bool) or not isinstance(valor, (int, float)):
        print("Tipo de resposta: informe um número, não texto ou True/False.")
        return False
    if not math.isfinite(valor) or not math.isclose(valor, esperado, abs_tol=tolerancia, rel_tol=1e-9):
        print("Cálculo: resultado diferente do esperado. Revise os dados e o denominador.")
        return False
    print("Resultado conferido. " + explicacao)
    return True
```

```python
import pandas as pd
df = pd.DataFrame({"id": [101, 102, 103], "turma": ["A", "A", "B"], "faltas": [0, 2, 1]})
print(df["turma"].value_counts())
print(df["faltas"].mean())
```

A aparece duas vezes e B uma. A média de faltas é 1. O ID permite identificar registros; sua média não informa aprendizagem.
