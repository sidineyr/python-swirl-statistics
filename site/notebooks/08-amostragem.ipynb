## Conceitos

População é o conjunto sobre o qual queremos concluir; amostra é o conjunto observado. Um processo de seleção determina quem pode entrar. Uma amostra grande por conveniência pode excluir grupos sistematicamente e continuar enviesada.

Variabilidade amostral é a mudança de uma estatística entre amostras obtidas pelo mesmo desenho. Viés é um desvio sistemático relacionado ao procedimento. Sorteios repetidos ajudam a estudar variabilidade; aumentar n não corrige automaticamente viés de seleção.

Desvio padrão descreve a dispersão dos valores individuais. Erro padrão descreve a variabilidade de um estimador, como a média, entre amostras. Sob hipóteses de observações independentes de uma população com variância finita, s / sqrt(n) estima o erro padrão da média. Não é uma fórmula universal para qualquer desenho.

A simulação usa uma população sintética conhecida e uma semente para permitir reprodução. A dispersão das médias sorteadas aproxima o erro padrão nesse experimento. Amostragem sem reposição de uma população finita pode exigir correção; aqui usamos reposição para ilustrar o caso independente. Não estamos construindo um intervalo de confiança neste módulo.

## Exemplo resolvido
Antes de executar, preveja a saída.

```python
from random import Random
from statistics import mean, pstdev
rng = Random(42)
populacao = list(range(10, 110))
medias = [mean(rng.choices(populacao, k=5)) for _ in range(200)]
print(mean(populacao))
print(mean(medias), pstdev(medias))
```

A média populacional é 59.5. As 200 médias variam; sua média tende a ficar perto de 59.5. A dispersão delas descreve o estimador, não a dispersão dos cem valores individuais.
