Quem estuda mais horas tende a ter nota maior nestes registros. As horas causaram o aumento?

## Conceitos

Correlação de Pearson mede associação linear entre duas variáveis quantitativas. Vai de −1 a 1: o sinal indica direção e o módulo descreve intensidade linear. Correlação zero não exclui uma relação não linear; uma variável constante torna o coeficiente indefinido.

Um gráfico de dispersão representa cada par (horas, nota). Observe forma, extremos e grupos antes de resumir por um coeficiente. Um único extremo pode alterar muito a correlação. Pares com valores ausentes são excluídos no cálculo usado aqui: declare quais pares entraram.

Associação não identifica uma causa. Conhecimento prévio, acesso a recursos ou outras características podem influenciar horas e nota. A direção temporal e o desenho de estudo importam. Para sustentar causalidade, precisamos de pressupostos e evidências adicionais.

Os dados deste curso foram inventados para ensinar. Mesmo uma correlação forte neles não é evidência sobre estudantes reais nem valida uma intervenção educacional.

```python
def conferir(valor, esperado, explicacao, tolerancia=0.01):
    """Aceita qualquer procedimento que produza um resultado numérico equivalente."""
    if isinstance(valor, bool) or not isinstance(valor, (int, float)):
        print("Tipo de resposta: informe um número, não texto ou True/False.")
        return False
    if not math.isfinite(valor) or not math.isclose(valor, esperado, abs_tol=tolerancia, rel_tol=1e-9):
        print("Cálculo: resultado diferente do esperado. Revise os dados e o denominador.")
        return False
    print("Resultado conferido. " + explicacao)
    return True
```

## Exemplo resolvido
Antes de executar, preveja a saída.

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.DataFrame({"horas": [1,2,3,4], "nota": [2,4,6,8]})
print(df["horas"].corr(df["nota"]))
df.plot.scatter(x="horas", y="nota", title="Pares sintéticos")
plt.show()
```

O coeficiente é 1: os quatro pontos seguem uma reta crescente. Isso descreve perfeitamente estes pares, mas não prova que acrescentar uma hora causaria dois pontos a mais.
