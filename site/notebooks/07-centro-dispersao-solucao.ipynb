Duas turmas têm notas médias iguais. Isso significa que seus resultados são semelhantes?

## Conceitos

Média aritmética considera todos os valores. Mediana divide os dados ordenados em duas metades; com quantidade par, é a média dos dois valores centrais. Um valor extremo afeta fortemente a média, mas pode não mudar a mediana.

Quantis localizam posições na distribuição. Q1, mediana e Q3 correspondem aos percentis 25, 50 e 75. O intervalo interquartil Q3 − Q1 descreve o espalhamento central. Existem convenções diferentes de interpolação; aqui usamos pandas com interpolação linear.

Amplitude é máximo menos mínimo. Desvio padrão resume distâncias em torno da média, nas mesmas unidades dos dados. Para descrever toda a população observada, a variância usa divisor n; para estimar variância populacional por uma amostra, usa n − 1. statistics.pstdev usa n, statistics.stdev usa n − 1 e pandas.std usa n − 1 por padrão.

Centro não substitui distribuição. Uma turma com notas [6,7,7,7,8] e outra com [3,5,7,10,10] têm média 7, mas amplitudes e dispersões diferentes. Com poucos casos, mostre os valores e evite conclusões sobre todas as turmas.

## Exemplo resolvido
Antes de executar, preveja a saída.

```python
from statistics import mean, median, pstdev, stdev
a = [6, 7, 7, 7, 8]
b = [3, 5, 7, 10, 10]
print(mean(a), mean(b))
print(max(a)-min(a), max(b)-min(b))
print(pstdev(a), pstdev(b))
print(stdev(a))
```

As médias são 7; amplitudes 2 e 7; desvios populacionais aproximadamente 0.632 e 2.757. O desvio amostral de A é aproximadamente 0.707. Mesma média não significa mesma dispersão.
