## Conceitos

Uma variável de programação é um nome associado a um objeto: tempos = [10, 20, 30] associa o nome tempos a uma lista. Uma variável estatística é uma característica observada, como tempo de estudo; cada estudante tem um valor dessa característica. Um nome Python pode guardar todos esses valores.

int representa inteiros, float números com parte decimal, str texto e bool valores lógicos. "10" é texto, enquanto 10 é número. int("10") converte esse texto em inteiro. Converter um identificador como "0012" para número pode perder informação e não é sempre apropriado.

Uma lista é uma sequência ordenada entre colchetes. A primeira posição tem índice 0: tempos[0]. len(tempos) conta elementos e sum(tempos) soma números. A média aritmética divide a soma pela quantidade; a lista precisa ter pelo menos um elemento.

```python
import math
from statistics import mean, median, pstdev, stdev
```

```python
def conferir(valor, esperado, explicacao, tolerancia=0.01):
    """Aceita qualquer procedimento que produza um resultado numérico equivalente."""
    if isinstance(valor, bool) or not isinstance(valor, (int, float)):
        print("Tipo de resposta: informe um número, não texto ou True/False.")
        return False
    if not math.isfinite(valor) or not math.isclose(valor, esperado, abs_tol=tolerancia, rel_tol=1e-9):
        print("Cálculo: resultado diferente do esperado. Revise os dados e o denominador.")
        return False
    print("Resultado conferido. " + explicacao)
    return True
```
