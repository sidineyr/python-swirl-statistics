Uma tabela de estudantes contém turma, horas de estudo e nota. Como selecionar as linhas da turma A?

## Conceitos

pandas é uma biblioteca para trabalhar com dados tabulares. Um DataFrame possui linhas, colunas e um índice. Neste curso cada linha representa um estudante sintético; cada coluna representa uma característica. Uma Series representa uma coluna ou sequência indexada.

CSV é um arquivo de texto com campos separados por delimitadores. pd.read_csv lê o arquivo; separador e codificação precisam corresponder ao arquivo. head() permite olhar as primeiras linhas; shape informa (linhas, colunas); dtypes informa tipos usados pelo software, que não substituem a classificação estatística.

df["turma"] seleciona uma coluna. df["turma"] == "A" produz uma máscara booleana; df.loc[mascara] seleciona linhas verdadeiras. Este filtro muda o grupo analisado: uma média da turma A não é automaticamente a média de toda a tabela.

```python
def conferir(valor, esperado, explicacao, tolerancia=0.01):
    """Aceita qualquer procedimento que produza um resultado numérico equivalente."""
    if isinstance(valor, bool) or not isinstance(valor, (int, float)):
        print("Tipo de resposta: informe um número, não texto ou True/False.")
        return False
    if not math.isfinite(valor) or not math.isclose(valor, esperado, abs_tol=tolerancia, rel_tol=1e-9):
        print("Cálculo: resultado diferente do esperado. Revise os dados e o denominador.")
        return False
    print("Resultado conferido. " + explicacao)
    return True
```
